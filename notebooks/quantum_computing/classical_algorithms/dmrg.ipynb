# Introduction

The Density Matrix Renormalization Group (DMRG) is a powerful classical method for computing ground states of one-dimensional quantum many-body systems by variationally optimizing a Matrix Product State (MPS) ansatz, enabled by the area-law entanglement structure of typical 1D ground states. One example is the XXX Heisenberg model Hamiltonian for a system of three interacting spin-1/2 particles ordered along a line (1D chain).

- **Classical Benchmark**: DMRG is also used as a classical gold standard for ground-state energies when benchmarking near-term quantum algorithms (e.g., VQE).

- **Limitation**: DMRG is highly efficient for 1D systems (low entanglement) thanks to the area law, which keeps required bond dimensions small. In contrast, when mapping 2D systems to 1D MPS (e.g., via cylinder geometry), bond dimensions typically grow exponentially with the system's width due to increasing entanglement across the cut.
