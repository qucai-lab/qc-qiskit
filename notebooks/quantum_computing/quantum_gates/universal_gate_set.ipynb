# The Universal Gate Set

A universal gate set is such that any possible gate operation can be reduced to a finite sequence of gates from the set.

- The `continuous gate set` $\{R_x(\theta), R_y(\theta), R_z(\theta),P(\lambda), CNOT\}$ comprising of single-qubit standard Pauli rotation gates, the single-qubit [phase shift gate $P(\lambda)$](https://en.wikipedia.org/wiki/Quantum_logic_gate#Phase_shift_gates) and the two-qubit $CNOT$ gate is a common universal set of quantum gates. `However, they are not easy to implement on encoded qubits for fault-tolerance.`

- The "Clifford+T" `finite gate set` denoted $\{H, S, CNOT, T\}$ is another widely used universal gate set (see Ref. [1], Chap. 10 for fault-tolerant constructions of those gates).

# Fault-Tolerant Implementations

Recall that fault-torelant implementation of Clifford gates is realized at the physical level using native gates inherent to the hardware architecture. Conversely, non-Clifford gates, such as the $T$ gate, require purification using magic state distillation **(see ref. [2]**).

Common physical implementations of native gates are:

1. {[RZ](https://quantumai.google/reference/python/cirq/rz), [Phased-X](https://quantumai.google/reference/python/cirq/PhasedXPowGate), [fSim](https://quantumai.google/reference/python/cirq/FSimGate)} is a native gate set of `Google's Sycamore` **(see ref. [3] and [4])**.

2. {[RZ](https://quantumai.google/reference/python/cirq/rz), [Phased-X](https://quantumai.google/reference/python/cirq/PhasedXPowGate), [CZ](https://quantumai.google/reference/python/cirq/CZ)} is a native gate set of `Google’s Bristlecone` **(see ref. [3] and [4])**.

3. {[ECR](https://docs.quantum.ibm.com/api/qiskit/qiskit.circuit.library.ECRGate), [RZ](https://docs.quantum.ibm.com/api/qiskit/qiskit.circuit.library.RZGate), [SX](https://docs.quantum.ibm.com/api/qiskit/qiskit.circuit.library.SXGate), [X](https://docs.quantum.ibm.com/api/qiskit/qiskit.circuit.library.XGate)} is a native gate set available to `IBM Brisbane, IBM Kyiv`, and `IBM Sherbrooke` quantum processors **(see ref. [5])**.

4. $\{X, SX, RX, CZ, RZ, RZZ \}$ is a native gate set supported by the [Heron QPU architecture](https://quantum.cloud.ibm.com/docs/en/guides/processor-types#heron), including processors such as `IBM Kingston`.

- $R_z$ and $R_x(\pi/2)$ can generate any arbitrary gate from $SU(2)$ **(see ref. [1])**.
