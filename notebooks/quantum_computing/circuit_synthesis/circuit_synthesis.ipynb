# Introduction

Unitary/Circuit synthesis start from a unitary matrix (or a high-level operation) and decomposes it into a sequence of quantum gates.

# Controlled-unitary Decomposition

See [Barenco et al.](https://journals.aps.org/pra/abstract/10.1103/PhysRevA.52.3457) Lemma 5.1 on page 10 for a Proof.

A Control-W gate can be decomposed into:

\begin{equation}
    C(W) = \Phi_1(\theta)A_2C(X)B_2C(X)C_2 \in SU(2),
\end{equation}

for $A$, $B$, and $C \in SU(2)$, where

\begin{equation}
\Phi_1 = \begin{pmatrix} 1 & 0 \\ 0 & e^{i\theta} \end{pmatrix} \otimes I_{2x2},
\end{equation}

is a phase gate applied to the first qubit. And $A_2$, $B_2$ and $C_2$ are gates applied to the second qubit.

See [Barenco et al.](https://journals.aps.org/pra/abstract/10.1103/PhysRevA.52.3457) Lemma 5.5 on page 13 for a Proof.

\begin{equation}
C(V) = A_2 C(X) B_2 \in SU(2),
\end{equation}

Where $V = R_z(\alpha) \cdot R_y(\theta) \cdot R_z(\alpha) \cdot \sigma_x$, with real values $\alpha$ and $\theta$.

See [Barenco et al.](https://journals.aps.org/pra/abstract/10.1103/PhysRevA.52.3457) Lemma 6.1 on page 14 for a Proof.

A $CCU$ gate can be decomposed into:

\begin{equation}
CCU \in SU(2^3) = CV(1,2)CX(0,1)CV^{\dagger}(1,2)CX(0,1)CV(0,2) \in SU(2),
\end{equation}

# KAK Decomposition

KAK decomposition [2] is a Lie-group factorization that arises from the Cartan decomposition of its underlying Lie algebra.

Recall that the Lie group $\mathfrak{G}$ is generated by the exponential map of its Lie algebra $\mathfrak{g}$:

\begin{equation}
    \mathfrak{G} = \exp(\mathfrak{g}).
\end{equation}

A Cartan decomposition is fundamentally a direct sum of a Lie subalgebra $\mathfrak{k}$ and a vector subspace $\mathfrak{p}$:

\begin{equation}
    \mathfrak{g} = \mathfrak{k} \oplus \mathfrak{p},
\end{equation}

satisfying the following commutation relations:

\begin{equation}
    [\mathfrak{k}, \mathfrak{k}] \subseteq \mathfrak{k}, \quad [\mathfrak{k}, \mathfrak{p}] \subseteq \mathfrak{p}, \quad [\mathfrak{p}, \mathfrak{p}] \subseteq \mathfrak{k}.
\end{equation}

The KAK decomposition then translates this algebraic structure into a factorization of elements of the corresponding Lie group $\mathfrak{G}$. For $G \in \mathfrak{G}$, the general expression for KAK decomposition is:

\begin{equation}
    G = K_1 A K_2,
\end{equation}

where $K_1, K_2 \in K$, with $K$ the Lie subgroup corresponding to the subalgebra $\mathfrak{k}$, and $A = \exp(\mathfrak{a})$ is an element of the abelian subgroup generated by the maximal abelian subspace $\mathfrak{a} \subseteq \mathfrak{p}$.

For a two-qubit Unitary gate, a KAK (Weyl) Decomposition is:

\begin{equation}
    U = (K_1^l \otimes K_1^r) e^{ik_x X \otimes X + ik_y Y \otimes Y + ik_z Z \otimes Z} (K_2^l \otimes K_2^r),
\end{equation}

where $U \in U(4)$, $K_1^l, K_1^r, K_2^l, K_2^r \in SU(2)$ and $\pi/4 \geq k_x \geq k_y \geq k_z$ (Weyl chamber).

# Cosine-Sine Decomposition (CSD)

CSD is a numerical matrix factorization method that uses a generalized type-AIII Cartan decomposition.

In Python, the CSD can be computed using the [scipy.linalg.cossin](https://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.cossin.html) function.

# Quantum Shannon Decomposition (QSD)

QSD [3, 4] is a recursive unitary matrix decomposition that uses the CSD as a key decomposition step. 

Attributes of QSD:

- Maintains unitarity throughout the decomposition.
- exploits the block structure introduced by the CSD at each recursive level.

The resulting factors can be implemented as:

- Multiplexed/uniformly controlled rotations.
- Entangling blocks.

# Block ZXZ Decomposition

Block-ZXZ decomposition [5, 6] is a recursive block matrix decomposition that uses type-AIII Cartan decompositions and type-A Cartan decompositions.
