# Introduction

Quantum phase estimation (QPE) [1] is a subroutine used to estimate the phase $\theta$ associated with the eigenvalue $e^{i 2\pi\theta}$ of a unitary operator $\hat{U}$ given a trial eigenstate $|\psi⟩$ with some overlap with the true eigenstate, without having to perform matrix diagonalization. 

Practical implementations require fault-tolerant quantum hardware with enough coherence times and gate fidelities capable of running deep quantum circuits, since the precision of QPE improves exponentially with the number of ancilla qubits $m$ in the first register:

\begin{equation}
    \epsilon \sim O(2^{-m}),
\end{equation}

where $\epsilon$ is the error tolerance. This is in contrast to the Variational Quantum Eigensolver (VQE), which runs on NISQ devices and requires $O(1/\epsilon^2)$ measurements to achieve the same precision. However, QPE has the potential to deliver higher precision than VQE.

# Applications

The algorithm is based on the quantum Fourier transform and is a key component in many quantum algorithms, including `Shor's algorithm` for factoring large integers and `Harrow–Hassidim–Lloyd (HHL) algorithm` for solving linear systems of equations.

# Implementation

The QPE algorithm can be broken down into the following steps:

1. **Initialization**: The algorithm starts with two quantum registers. The first register is initialized to the state $|0⟩^{\otimes m}$ with $m$ qubits, while the second register is prepared in an eigenstate $|\psi⟩$ of the unitary operator $\hat{U}$ to be analyzed.
   
2. **Superposition**: A Hadamard gate is applied to every qubit in the first register, creating a superposition of all possible states. These are the qubits that will be used to encode the phase information and that will be measured at the end of the algorithm.
   
3. **Controlled Unitary Operations**: A sequence of controlled-$\hat{U}^{2^m}$ operations is applied on the first register with target qubits on the second register, where $m = 0, 1, \ldots, n-1$ is the position of the $m$-th control qubit.
   
4. **Inverse Quantum Fourier Transform**: After the controlled-$\hat{U}^{2^m}$ operations, an inverse quantum Fourier transform is applied to the first register. This step is crucial for extracting the phase information from the superposition.
   
5. **Measurement**: Finally, the first register is measured, yielding an estimate of the phase $\theta$ in $\hat{U} |\psi\rangle = e^{i 2\pi\theta} |\psi\rangle$.
