# Introduction

Quantum phase estimation (QPE) relies on fault-tolerant quantum hardware because it typically involves deep circuits and demands long coherence times. This is in contrast to the Variational Quantum Eigensolver (VQE), which runs on NISQ devices. However, QPE has the potential to deliver higher precision than VQE.
