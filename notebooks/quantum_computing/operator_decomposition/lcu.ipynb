# Definition

Linear Combination of Unitaries (LCU) is a quantum algorithmic tool that decomposes a non-unitary operator as a weighted sum of unitary operators.

The general expression for a LCU is:

\begin{align}
    A := \sum_{k=0}^{n-1} a_k U_k.
\end{align}

where $a_k$ are complex coefficients and $U_k$ are unitary operators. Note that $A$ is not necessarily unitary, but it can be implemented on a quantum computer using the LCU technique.

# Applications

- Used to approximate a time Unitary evolution $\hat{U}=e^{-i\hat{H}t/\hbar}$ with $\hat{H} = \sum_k a_k U_k$ for [quantum Hamiltonian simulation](../quantum_algorithms/trotterization.ipynb).

- It provides a method for implementing sums of unitaries in a quantum circuit using ancilla qubits and controlled-unitary operations.
