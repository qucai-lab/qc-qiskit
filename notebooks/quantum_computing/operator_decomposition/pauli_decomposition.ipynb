# Definition

Since measurements in the quantum computer are performed in the computational basis a.k.a Pauli-Z basis, when the quantum observable $\hat{\mathcal{O}}$ is not diagonalized in the computational basis, a change of basis circuit is necessary to rotate from the eigenbasis of the observable into the computational basis:

$$\langle \psi| \hat{\mathcal{O}} |\psi\rangle = \langle 0^{\otimes n}|\hat{U}^{\dagger}W^{\dagger} \hat{\mathcal{O}} W\hat{U}|0^{\otimes n}\rangle \in \mathbb{R},$$

where $W$ denotes the change of basis circuit.

However, finding the change of basis circuit for a general observable is not a trivial task. Therefore, it is useful to write the Hermitian observable into the Pauli basis, i.e, decompose it into a sum of Pauli operators. 

The idea of a Pauli decomposition is that any $n$-qubit operator can be written as a linear combination of tensor products of Pauli operators:

$$\hat{\mathcal{O}} := \sum_{j=1}^{d=2^n} c_j P_j, \ \ \ P_j \in \{I, X, Y, Z\}^{\otimes n}.$$

The set $\{I, X, Y, Z\}$ forms a complete basis.

With that, the new measurement becomes:

$$\langle \psi| \hat{\mathcal{O}} |\psi\rangle =\langle \psi| \Big(\sum_{j=1}^{d=2^n} c_j P_j \Big) |\psi\rangle= \sum_{j=1}^{d=2^n} c_j \langle \psi | P_j | \psi \rangle \in \mathbb{R}.$$

# Pauli Coefficients

Note that the pauli operators $P_j$ are Hermitian (i.e., $P_j^\dagger = P_j$) and orthogonal with respect to the Hilbert-Schmidt inner product:

\begin{align}
    \text{Tr}(P_j P_k) = 2^n \delta_{jk},
\end{align}

where $2^n$ comes from $||P_j||^2_{HS} = \text{Tr}(P_j^\dagger P_j) = \text{Tr}(P_j P_j) = \text{Tr}(P_j^2) = \text{Tr}(I) = 2^n$.

The coefficients $c_j$ can be calculated using the definition of the [Hilbert-Schmidt inner product](../../algebra/inner_product.ipynb):

\begin{align}
    \langle P_j, \hat{\mathcal{O}} \rangle_{HS} &= \text{Tr}(P_j^\dagger \hat{\mathcal{O}}) \\
     &= \text{Tr}(P_j \hat{\mathcal{O}}) \\
     &= \text{Tr} \left(P_j \sum _{k=1}^{d=2^n} c_k P_k\right) \\
\end{align}

Applying the trace property $\text{Tr}(A+B) = \text{Tr}(A) + \text{Tr}(B)$ and $\text{Tr}(\alpha A) = \alpha \text{Tr}(A)$, we have:

\begin{align}
     &= \sum _{k=1}^{d=2^n} c_k \text{Tr}(P_j P_k) \\
     &= \sum _{k=1}^{d=2^n} c_k 2^n \delta_{jk} \\
    &= 2^n c_j.
\end{align}

Therefore:

\begin{align}
    c_j = \frac{1}{2^n} \text{Tr}(P_j \hat{\mathcal{O}}).
\end{align}
