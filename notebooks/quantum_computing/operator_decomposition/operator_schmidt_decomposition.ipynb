# Definition

An operator $A$ acting on a bipartite space $\mathcal{H}_A \otimes \mathcal{H}_B$ has an operator Schmidt decomposition of the form:

\begin{align}
    A := \sum_k s_k A_k \otimes B_k,
\end{align}

where $s_k \geq 0$ are the operator Schmidt coefficients, and $\{A_k\}$ and $\{B_k\}$ are orthonormal operator bases under the [Hilbert-Schmidt inner product](../../algebra/inner_product.ipynb), i.e., they satisfy the following orthonormality conditions:

\begin{align}
    \text{Tr}(A_j^\dagger A_k) = \delta_{jk}, \quad \text{Tr}(B_j^\dagger B_k) = \delta_{jk}.
\end{align}



# Applications

-  Used for quantifying nonlocality.

-  Used in tensor network methods.

# Examples

Consider the two-qubit unitary operator defined as:

\begin{align}
    e^{-i \frac{\theta}{2} Z \otimes Z} = cos\left( \frac{\theta}{2} \right) I \otimes I − i sin\left(\frac{\theta}{2}\right) Z \otimes Z.
\end{align}

For two qubits:

\begin{align}
    \text{Tr} (\mathbb{I}^\dagger \mathbb{I}) = 2, \quad \text{Tr} (Z^\dagger Z) = 2.
\end{align}

The normalized operators are:

\begin{align}
    A_0 = B_0 = \frac{I}{\sqrt{2}}, \quad A_1 = B_1 = \frac{Z}{\sqrt{2}}.
\end{align}

The Schmidt coefficients come from the Singular Value Decomposition (SVD):

\begin{align}
    M = U \Sigma V^\dagger = \sum_k \sigma_k |u_k\rangle \langle v_k|,
\end{align}

where $M$ is the matrix representation of the operator in the basis of $A_k \otimes B_k$, and $\Sigma$ is a diagonal matrix containing the Schmidt coefficients $\sigma_k$. For our example, $\Sigma$ has the form:

\begin{align}
    \Sigma = \begin{pmatrix}
    \sigma_0 & 0 \\
    0 & \sigma_1
    \end{pmatrix}.
\end{align}

The singular values of this matrix are: 

\begin{align} 
    \sigma_0 = 2 \left| \cos\left( \frac{\theta}{2} \right) \right|, \quad \sigma_1 = 2 \left| \sin\left( \frac{\theta}{2} \right) \right|. 
\end{align}
