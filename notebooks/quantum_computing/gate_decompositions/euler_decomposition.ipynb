# Euler Decomposition

Euler decomposition can express any single-qubit unitary operation, up to a global phase, as a product of three rotations about two orthogonal axes.

Note: Despite the following notations (e.g., Z-X-Z), this is not a local conjugation by Unitary transformation ($UPU^{\dagger}$).

## Z-X-Z Decomposition

Z-X-Z Decomposition for a single-qubit Unitary gate $U$:

\begin{equation}
    U = e^{i\alpha}R_z(\beta)R_x(\gamma)R_z(\delta).
\end{equation}

Where $\alpha$, $\beta$, $\gamma$ and $\delta$ are real-valued numbers.

Example:


\begin{equation}
    R_y(\theta) = R_z(\pi/2) R_x(\theta) R_z(-\pi/2)
\end{equation}

## Z-Y-Z Decomposition

See Theorem 4.1 of Ref. [1] on page 175 for a Proof.

Z-Y-Z Decomposition for a single-qubit Unitary gate $U$:

\begin{equation}
    U = e^{i\alpha}R_z(\beta)R_y(\gamma)R_z(\delta).
\end{equation}

Where $\alpha$, $\beta$, $\gamma$ and $\delta$ are real-valued numbers.

## X-Y-X Decomposition

See Exercise 4.10 of Ref. [1] on page 176.

X-Y-X Decomposition for a single-qubit Unitary gate $U$:

\begin{equation}
    U = e^{i\alpha} R_x(\beta)R_y(\gamma)R_x(\delta).
\end{equation}

Where $\alpha$, $\beta$, $\gamma$ and $\delta$ are real-valued numbers.

## X-Z-X Decomposition

X-Z-X Decomposition for a single-qubit Unitary gate $U$:

\begin{equation}
    U = e^{i\alpha}R_x(\beta)R_z(\gamma)R_x(\delta).
\end{equation}

Where $\alpha$, $\beta$, $\gamma$ and $\delta$ are real-valued numbers.

Example:

\begin{equation}
    R_y(\theta) = R_x(-\pi/2) R_z(\theta) R_x(\pi/2)
\end{equation}

## V-Z Decomposition

\begin{equation}
    U = e^{i\alpha}R_z(\beta)V^{\dagger}R_z(\gamma)VR_z(\delta).
\end{equation}

Where:

- $V=X^{1/2} = \sqrt(X)$.
- $R_z(\theta)=V^{\dagger}R_y(\theta)V$.

\begin{equation}
    CZ = exp(i(0 \cdot X \otimes X + 0 \cdot Y \otimes Y + \frac{\pi}{4} Z \otimes Z)).
\end{equation}

**Exercise A:** Suppose $\hat{m}$ and $\hat{n}$ are non-parallel real unit vectors in three dimensions. Use Theorem 4.1 to show that an arbitrary single qubit unitary U may be written

$$U=e^{i\alpha} R_{\hat{n}}(\beta)R_{\hat{m}}(\gamma)R_{\hat{n}}(\delta).$$

for appropriate choices of $\alpha$, $\beta$, $\gamma$ and $\delta$.

Tip: See exercise 4.11 of Ref. [1] on page 176.

**Exercise B:** Prove that a Unitary gate $U$ acting on a single qubit can always be decomposed into:

\begin{equation}
    U = e^{i\alpha}AXBXC,
\end{equation}

where $\alpha$ is a overall phase factor and $ABC=I$ for unitary operators $A$, $B$, and $C \in SU(2)$ acting on a single qubit.

Tip: See corollary 4.2 of Ref. [1] on page 176, or the proof of Lemma 4.3 from [Barenco et al.](https://journals.aps.org/pra/abstract/10.1103/PhysRevA.52.3457)
