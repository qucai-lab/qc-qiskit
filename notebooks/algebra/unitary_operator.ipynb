# Definition

A unitary matrix $\hat{U}$ is a complex square matrix that satisfies the following condition:

\begin{equation}
    \hat{U}^\dagger \hat{U} = \hat{U} \hat{U}^\dagger = \hat{I},
\end{equation}

where $\hat{U}^\dagger$ is the hermitian adjoint given by the conjugate transpose of $\hat{U}$, and $\hat{I}$ is the identity matrix.

Unitary matrices preserve the inner product, meaning that for any two vectors $|\psi⟩$ and $|\phi⟩$, we have:

\begin{equation}
    (\hat{U} |\psi\rangle)^{\dagger} (\hat{U} |\phi\rangle) = |\psi\rangle^{\dagger} \hat{U}^{\dagger} (\hat{U} |\phi\rangle) = \langle \psi| \hat{U}^{\dagger} \hat{U} |\phi\rangle = \langle\psi | \phi\rangle.
\end{equation}

# Examples

Examples of unitary matrices include the Pauli matrices, which are fundamental in quantum mechanics:

\begin{align}
X &:= 
\begin{pmatrix}
0 & 1 \\
1 & 0
\end{pmatrix}. \\
Y &:=
\begin{pmatrix}
0 & -i \\
i & 0
\end{pmatrix}. \\
Z &:= 
\begin{pmatrix}
1 & 0 \\
0 & -1
\end{pmatrix} .
\end{align}
