# Definition

A normal operator $\hat{A}$ is a linear operator that commutes with its hermitian adjoint $\hat{A}^\dagger$:

\begin{align}
    [\hat{A}, \hat{A}^\dagger] &= \hat{A}\hat{A}^\dagger - \hat{A}^\dagger\hat{A} = 0 \\
    &\implies \hat{A}\hat{A}^\dagger = \hat{A}^\dagger\hat{A}.
\end{align}

Normal operators have the property that they can be diagonalized by a unitary transformation, which means that there exists a unitary operator $\hat{U}$ such that:

\begin{equation}
    \hat{A} = \hat{U} \hat{D} \hat{U}^\dagger,
\end{equation}

where the diagonal entries of $\hat{D}$ contain the eigenvalues of $\hat{A}$, while the columns of $\hat{U}$ contain the corresponding eigenvectors.

# Examples

Examples of normal operators include unitary operators $\hat{U}$ and Hermitian operators $\hat{H}$.
