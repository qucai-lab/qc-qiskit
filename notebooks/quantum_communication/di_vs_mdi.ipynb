# DI vs MDI QKD <a name="di-vs-mdi-qkd"></a>

- Device-independent (DI) QKD:
    - Is fully device-independent. 
    - There is no trust in either the source or the measurement device.
    - Relies on the violation of Bell's inequalities.
    - The shared quantum state between Alice and Bob must be entangled.

- Measurement-device-independent (MDI) QKD:
    - There is trust in the source, but no trust in the measurement device.
    - Uses Bell state measurements.
    - Input states are separable (not entangled).
