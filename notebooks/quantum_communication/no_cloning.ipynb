# Theorem

The `no-cloning theorem` states that:

_There is no unitary operator $\hat{U}$ acting on $\mathcal{H}_A \otimes \mathcal{H}_B$ that can satisfy the following condition for all normalized states $|\psi \rangle_A \in \mathcal{H}_A$ and $|e \rangle_B \in \mathcal{H}_B$ with $\dim(\mathcal{H}_A) \simeq \dim(\mathcal{H}_B)$:_

\begin{equation}
\hat{U} \left( |\psi \rangle_A \otimes |e \rangle_B \right) = e^{i\alpha} |\psi \rangle_A \otimes |\psi \rangle_B.
\end{equation}

In other words, no single unitary operation can perfectly copy an arbitrary, unknown, and normalized quantum state $|\psi \rangle_A$ over some initial (blank) state $|e \rangle_B$ that is independent of $|\psi\rangle_A$.

Obs: Cloning is only possible for mutually orthogonal states.

# Proof

Given two arbitrary states $|\psi\rangle_A$ and $|\phi\rangle_A$ and an initial state $|e\rangle_B$:

\begin{align}
\hat{U} \left( |\psi \rangle_A \otimes |e \rangle_B \right) &= |\psi \rangle_A \otimes |\psi \rangle_B, \\
\hat{U} \left( |\phi \rangle_A \otimes |e \rangle_B \right) &= |\phi \rangle_A \otimes |\phi \rangle_B.
\end{align}

Since the initial state $|e\rangle$ is normalized, i.e., $\langle e|e \rangle = 1$, we can write the inner product between the two arbitrary states as:

\begin{align}
\langle \psi | \phi \rangle_A &= \langle \psi | \phi \rangle_A \otimes \langle e|e \rangle_B,
\end{align}

Then using the algebraic idendity $(A \otimes B)(C \otimes D) = (AC) \otimes (BD)$, we have:

\begin{align}
\langle \psi | \phi \rangle_A \otimes \langle e|e \rangle_B = (\langle \psi |_A \otimes \langle e |_B) (|\phi \rangle_A \otimes |e \rangle_B).
\end{align}

And since the operator $\hat{U}$ is unitary, it preserves the inner product:

\begin{align}
&= (\langle \psi |_A \otimes \langle e |_B) \hat{U}^\dagger \hat{U} (|\phi \rangle_A \otimes |e \rangle_B).
\end{align}

If we now apply the cloning operation on both sides, we obtain:

\begin{align}
&= (\langle \psi |_A \otimes \langle \psi |_B) (|\phi \rangle_A \otimes |\phi \rangle_B).
\end{align}

Which after using identity $(A \otimes B)(C \otimes D) = (AC) \otimes (BD)$, can be rewritten as:

\begin{align}
&= \langle \psi | \phi \rangle_A \otimes \langle \psi | \phi \rangle_B.
\end{align}

However, since the tensor product of two scalars is just a multiplication, we have:

\begin{align}
&=  \langle \psi | \phi \rangle_A \langle \psi | \phi \rangle_B \\
&= \langle \psi | \phi \rangle^2.   
\end{align}

Then, $\langle \psi | \phi \rangle = \langle \psi | \phi \rangle^2$ implies that $\langle \psi | \phi \rangle$ must be either 0 or 1, meaning that the states $|\psi\rangle_A$ and $|\phi\rangle_A$ are either orthogonal or identical, respectively.

Therefore, non-orthogonal states cannot be perfectly cloned.

# Implications

1. **Quantum Teleportation:** The no-cloning theorem ensures that the original state is destroyed during the teleportation process, preventing the existence of two identical copies of the same quantum state.

2. **Quantum Cryptography:** The no-cloning theorem is a fundamental principle underlying the security of quantum cryptographic protocols, such as Quantum Key Distribution (QKD). It ensures that an eavesdropper cannot make perfect copies of quantum states used for secure communication, thus preventing undetected interception of information.

# Example



Consider two orthogonal states $|0\rangle$ and $|1\rangle$, and initial state $|e\rangle = |0\rangle$. A unitary operator $\hat{U}$ can be defined such that:

\begin{align}
\hat{U} (|0\rangle \otimes |0\rangle) &= |0\rangle \otimes |0\rangle, \\
\hat{U} (|1\rangle \otimes |0\rangle) &= |1\rangle \otimes |1\rangle.
\end{align} 



One such unitary operator is the Controlled-NOT (CNOT) gate, which flips the state of the second qubit (target) if the first qubit (control) is in the state $|1\rangle$:

\begin{align}
\text{CNOT} (|0\rangle \otimes |0\rangle) &= |0\rangle \otimes |0\rangle, \\
\text{CNOT} (|1\rangle \otimes |0\rangle) &= |1\rangle \otimes |1\rangle.
\end{align}

However, for an arbitrary state $|\psi\rangle = \alpha |0\rangle + \beta |1\rangle$ with $|\alpha|^2 + |\beta|^2 = 1$, applying the CNOT gate yields:

\begin{align}
\text{CNOT} (|\psi\rangle \otimes |0\rangle) &= \text{CNOT} \Big( (\alpha |0\rangle + \beta |1\rangle) \otimes |0\rangle \Big) \\
&= \alpha \text{CNOT} (|0\rangle \otimes |0\rangle) + \beta \text{CNOT} (|1\rangle \otimes |0\rangle) \\
&= \alpha |0\rangle \otimes |0\rangle + \beta |1\rangle \otimes |1\rangle \\
&\neq |\psi\rangle \otimes |\psi\rangle.
\end{align}
