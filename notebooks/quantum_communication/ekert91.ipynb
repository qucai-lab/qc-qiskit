# Theory

Ekert 91 (E91) is an entanglement-based and discrete-variable quantum key distribution (QKD) protocol proposed by Artur Ekert in 1991. It utilizes the principles of quantum entanglement and Bell's theorem to ensure secure communication between two parties, commonly referred to as Alice and Bob. 

In the E91 protocol, a source generates pairs of entangled qubits (usually in the singlet state) and sends one qubit from each pair to Alice and the other to Bob. Both parties perform measurements on their respective qubits using randomly chosen bases. The measurement outcomes are then used to generate a shared secret key.

Entanglement-based QKD protocols rely on the monogamy of entanglement to ensure that if Alice and Bob share strong quantum correlations, no third party (eavesdropper) can be entangled with them in the same way, i.e, being able to share the same correlations without being detected. Security is also supported by the no-cloning theorem.
