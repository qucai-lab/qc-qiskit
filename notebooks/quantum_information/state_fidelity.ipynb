# Definition

In noisy environments, the evolved quantum state must be represented by a mixed density operator. The [state fidelity](https://en.wikipedia.org/wiki/Fidelity_of_quantum_states) between two mixed density operators (Hermitian, positive-semidefinite, and unit trace matrices) $\rho$ and $\sigma$ is:

\begin{align}
F(\rho,\sigma) = \left(tr \sqrt{\sqrt{\rho} \sigma \sqrt{\rho}}\right)^2,
\end{align}

with $\rho= \sum_jp_j |\rho_j\rangle \langle \rho_k|$ and $\sigma=\sum_k p_k|\sigma_k\rangle \langle \sigma_k|$, where $p_j$ and $p_k$ are the corresponding probabilities from a marginal probability distribution, and also the eigenvalues according to the spectral decomposition theorem.

If one of the states is a pure state (e.g. $\rho=|\psi_{\rho}\rangle \langle \psi_{\rho}|$), the equation reduces to the fidelity between a pure and a mixed state:

\begin{align}
F(\sigma, |\psi\rangle) = \langle\psi_{\rho}|\sigma|\psi_{\rho}\rangle = \sum_{j,k=0}^{d-1}\psi_{j}^{*}\sigma_{j,k}\psi_{k}.
\end{align}

If both states are pure states, i.e, have pure density matrices $\rho=|\psi_{\rho}\rangle \langle \psi_{\rho}|$ and $\sigma=|\psi_{\sigma}\rangle \langle \psi_{\sigma}|$, then

\begin{align}
F(\rho, \sigma) = |\langle\psi_{\rho}|\psi_{\sigma}\rangle|^2.
\end{align}
